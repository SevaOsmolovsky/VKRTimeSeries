```python
# изначальный

"""
    Модель PatchTST для многомерного прогнозирования временных рядов.

    Основные этапы:
      1. Входной тензор размерности [batch, n_vars, seq_len].
      2. Реализация channel‑independence: объединение batch и каналов.
      3. Разбиение каждого ряда на перекрывающиеся патчи.
      4. Линейное преобразование патчей в эмбеддинги размерности d_model.
      5. Добавление позиционного кодирования (обучаемое).
      6. Пропускание последовательности патчей через трансформерный энкодер.
      7. Сплющивание закодированных патчей и преобразование в прогноз.
"""

class PatchTST(nn.Module):

    # конструктор определяет, из каких слоев состоит модель
    # (3 обязательных параметра)
    def __init__(self, seq_len, pred_len, n_vars, patch_len=16, stride=8, d_model=64, n_heads=8, e_layers=2, dropout=0.2):
        """
          ! seq_len   : длина окна
          ! pred_len  : длина прогноза
          ! n_vars    : число признаков (7 вместе с целевой переменной 'OT')
            patch_len : длина одного патча
            stride    : шаг при нарезке патчей
          ? d_model   : размерность скрытого представления (внутренняя размерность)
          ? n_heads   : количество голов в multi-head attention
          ? e_layers  : число слоёв трансформерного энкодера
          ? dropout   : вероятность dropout для регуляризации
        """
        super(PatchTST, self).__init__()

        # параметры для forward
        self.seq_len = seq_len
        self.pred_len = pred_len
        self.n_vars = n_vars
        self.patch_len = patch_len
        self.stride = stride
        self.d_model = d_model

        # кол-во патчей в окне
        # пример для окна 96: 11 патчей по 16 точек со сдвигом 8
        self.n_patches = int((seq_len - patch_len) / stride) + 1

        # патч должен преобразовываться в вектор размерности d_model (эмбеддинг патча)
        # матрица весов размером (d_model, patch_len) инициализируется случайными числами
        # Вектор смещений (bias) размером (d_model,) инициализируется случайно
        self.patch_embed = nn.Linear(patch_len, d_model)

        # ????????????????????????????
        # создание обучаемого параметра для позиционного кодирования патчей
        # добавляется к каждому эмбеддингу патча
        # пример размерности - (1, 11, 64)
        self.pos_enc = nn.Parameter(torch.randn(1, self.n_patches, d_model))

        # ????????????????????????????
        # стандартизация эмбеддинга 
        self.norm = nn.LayerNorm(d_model)
        
        # ?????????????????????????????
        # трансформерный энкодер: обрабатывает последовательности патчей
        encoder_layer = nn.TransformerEncoderLayer(
            d_model=d_model,
            nhead=n_heads, # связь между патчами ???
            dim_feedforward=d_model * 4, # расширение в FFN
            dropout=dropout,
            activation='gelu',         # активация GELU (плавная альтернатива ReLU)
            batch_first=True  # размерность (batch, seq, features)
        )
        self.transformer = nn.TransformerEncoder(encoder_layer, num_layers=e_layers) # создает стек из e_layers таких слоев; выход предыдущего слоя подается на вход следующему

        # патчи преобразуются в прогноз-вектор длины pred_len
        self.head = nn.Linear(self.n_patches * d_model, pred_len)

    def forward(self, x):
        """
        Прямой проход модели.

        Вход:
            x : torch.Tensor формы (batch_size, n_vars, seq_len)
                - batch_size : число примеров в батче
                - n_vars     : число переменных (каналов)
                - seq_len    : длина входной последовательности

        Выход:
            pred : torch.Tensor формы (batch_size, pred_len, n_vars)
                - прогноз для каждой переменной на pred_len шагов вперёд
        """
        batch_size = x.shape[0]   # размер батча

        # исходная размерность: (batch, n_vars, seq_len)
        # новая размерность: (batch, seq_len, n_vars)
        # для реализации channel‑independence объедининяются batch и n_vars
        x = x.permute(0, 2, 1)
        
        # объединение batch и n_vars в одну размерность
        # теперь каждый признак обрабатывается как отдельный пример
        # все каналы используют общие веса трансформера, но каждый канал идёт по своему пути
        x = x.reshape(batch_size * self.n_vars, self.seq_len)
        
        # нарезка скользящих окон
        # Вход: (batch*n_vars, seq_len)
        # Выход: (batch*n_vars, n_patches, patch_len)
        # Каждый патч — это локальный фрагмент временного ряда.
        patches = x.unfold(dimension=1, size=self.patch_len, step=self.stride)
        
        # преобразование каждого патча (длины patch_len) в d_model-мерный вектор.
        # выход: (batch*n_vars, n_patches, d_model)
        patches = self.patch_embed(patches)

        
        # pos_enc складывается с каждым элементом батча.
        # это позволяет трансформеру различать порядок патчей.
        patches = patches + self.pos_enc

        # стандартизация
        patches = self.norm(patches)

        # Вход: (batch*n_vars, n_patches, d_model)
        # Выход: той же размерности — закодированные патчи.
        # вычисляется между патчами (по временной оси) внутри каждого канала.
        encoded = self.transformer(patches)

        # формирование прогноза
        # преобразование всех патчей в один вектор для каждого канала
        # из (batch*n_vars, n_patches, d_model) -> (batch*n_vars, n_patches * d_model)
        encoded_flat = encoded.reshape(batch_size * self.n_vars, -1)

        # Выходной слой выдаёт прогноз на pred_len шагов для каждого канала:
        # (batch*n_vars, pred_len)
        pred = self.head(encoded_flat)

        # ========== Шаг 9. Восстановление исходной размерности ==========
        # Разделяем обратно batch и каналы:
        # (batch*n_vars, pred_len) -> (batch, n_vars, pred_len)
        pred = pred.reshape(batch_size, self.n_vars, self.pred_len)

        # Меняем оси, чтобы получить (batch, pred_len, n_vars) — это стандартный формат
        # для сравнения с целевыми значениями в loss-функции (например, MSE).
        pred = pred.permute(0, 2, 1)

        return pred
    

seq_len = 96 # длина последовательности
pred_len = 96 # длина прогноза
patch_len = 16 # длина патча
stride = 8 # шаг при нарезке
d_model = 64 # размер эмбеддинга
n_heads = 8 # ??
e_layers = 2 # кол-во слоев в трансформере
dropout = 0.2 # доля отключенных нейронов
batch_size = 8 # ??
epochs = 50 # кол-во эпох в обучении
lr = 1e-4 # скорость обучения
patience = 7 # ??
weight_decay = 1e-5 # ??
train_ratio = 0.7 # доля данных для обучения
val_ratio = 0.15 # доля данных для валидации
dataset = './ETTh1.csv' 


# диапазон данных: 2016-07-01 00:00:00 -> 2018-06-26 19:00:00
# преобразование столбца даты в индекс
df = pd.read_csv(dataset, parse_dates=['date'], index_col='date')
features = ['HUFL', 'HULL', 'MUFL', 'MULL', 'LUFL', 'LULL', 'OT']
data = df[features].values
total_len = len(data)
print(f"Размер данных: {total_len}")


# разделение данных на обучение, валидацию, тестирование
train_border = int(total_len * train_ratio)
val_border = int(total_len * (train_ratio + val_ratio))

train_data = data[:train_border]
val_data = data[train_border:val_border]
test_data = data[val_border:]


# нормализация
scaler = StandardScaler()
train_scaled = scaler.fit_transform(train_data)
val_scaled = scaler.transform(val_data)
test_scaled = scaler.transform(test_data)


# преобразование в тензоры и создание окон
def create_sequences(data, seq_len, pred_len):
    xs, ys = [], []
    for i in range(len(data) - seq_len - pred_len + 1):
        x = data[i:i + seq_len] # 96 временных меток
        y = data[i + seq_len:i + seq_len + pred_len] # 96 временных меток
        xs.append(x)
        ys.append(y)
    return np.array(xs), np.array(ys)

# трехмерные массивы размерности (windows, seq_len, n_vars)
X_train, y_train = create_sequences(train_scaled, seq_len, pred_len)
X_val, y_val = create_sequences(val_scaled, seq_len, pred_len)
X_test, y_test = create_sequences(test_scaled, seq_len, pred_len)


# объединяет два тензора X и y в один объект
# при обращении по индексу возвращается пара (X[i], y[i]), где их размерность (96, 7)
train_dataset = TensorDataset(torch.FloatTensor(X_train), torch.FloatTensor(y_train))
val_dataset = TensorDataset(torch.FloatTensor(X_val), torch.FloatTensor(y_val))
test_dataset = TensorDataset(torch.FloatTensor(X_test), torch.FloatTensor(y_test))


# разбивка датасета на 1500 батчей (8 окон в каждом батче) + перемешивание в обучении, чтобы не запоминать порядок окон
# пример: 12000 / 8 = 1500 батчей
train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=batch_size, shuffle=False)
test_loader = DataLoader(test_dataset, batch_size=batch_size, shuffle=False)



n_vars = len(features) # 7 вместе с целевой переменной
target_idx = features.index('OT') # целевая переменная



# создание модели
model = PatchTST(
    seq_len=seq_len, pred_len=pred_len, n_vars=n_vars,
    patch_len=patch_len, stride=stride,
    d_model=d_model, n_heads=n_heads, e_layers=e_layers,
    dropout=dropout
).to(device) #  перемещает все тензоры и слои модели на GPU

# Adam – адаптивный оптимизатор, автоматически подстраивает скорость обучения
# model.parameters() – передаёт в оптимизатор все обучаемые параметры модели: веса и смещения всех слоёв (patch_embed, transformer, head, pos_enc, norm и т.д.). Эти параметры будут обновляться в процессе обучения.
# weight_decay – коэффициент L2-регуляризации (штраф за большие веса), добавляет к функции потерь слагаемое weight_decay * sum(weight^2), что помогает предотвратить переобучение, заставляя веса оставаться небольшими. В вашем случае weight_decay=1e-5
optimizer = torch.optim.Adam(model.parameters(), lr=lr, weight_decay=weight_decay)
criterion = nn.MSELoss() # среднеквадратичная ошибка

# ???
scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer, mode='min', factor=0.5, patience=5, verbose=True
)



print("Обучение: \n")
best_val_loss = float('inf') # переменную для хранения наилучшей валидационной потери
counter = 0 # счетчик эпох

for epoch in range(1, epochs + 1):
    # Train
    model.train() # переключает модель в режим обучения
    train_loss = 0.0 # обнуление накопителя суммы потерь за эпоху
    for X_batch, y_batch in train_loader:
        X_batch, y_batch = X_batch.to(device), y_batch.to(device)
        X_batch = X_batch.permute(0, 2, 1)  # [batch, seq_len, n_vars] -> [batch, n_vars, seq_len]
        optimizer.zero_grad()
        pred = model(X_batch)
        loss = criterion(pred, y_batch)
        loss.backward()
        optimizer.step()
        train_loss += loss.item()
    train_loss /= len(train_loader)

    # Validation
    model.eval()
    val_loss = 0.0
    with torch.no_grad():
        for X_batch, y_batch in val_loader:
            X_batch, y_batch = X_batch.to(device), y_batch.to(device)
            X_batch = X_batch.permute(0, 2, 1)
            pred = model(X_batch)
            loss = criterion(pred, y_batch)
            val_loss += loss.item()
    val_loss /= len(val_loader)

    scheduler.step(val_loss)

    if val_loss < best_val_loss:
        best_val_loss = val_loss
        counter = 0
        torch.save(model.state_dict(), 'best_model.pth')
    else:
        counter += 1
        if counter >= patience:
            print(f"Ранняя остановка на эпохе {epoch}")
            break

    if epoch % 10 == 0 or epoch == 1:
        print(f"Эпоха {epoch:3d}/{epochs} | Train Loss: {train_loss:.6f} | Val Loss: {val_loss:.6f}")

print("Обучение завершено. Загружаем лучшую модель для тестирования...")
model.load_state_dict(torch.load('best_model.pth', weights_only=False))


model.eval()
all_x, all_y, all_pred = [], [], []
with torch.no_grad():
    for X_batch, y_batch in test_loader:
        X_batch, y_batch = X_batch.to(device), y_batch.to(device)
        X_batch = X_batch.permute(0, 2, 1)
        pred = model(X_batch)
        all_x.append(X_batch.cpu().numpy())
        all_y.append(y_batch.cpu().numpy())
        all_pred.append(pred.cpu().numpy())

all_x = np.concatenate(all_x, axis=0)
all_y = np.concatenate(all_y, axis=0)
all_pred = np.concatenate(all_pred, axis=0)

# Метрики в исходном масштабе
mean_ot = scaler.mean_[target_idx]
scale_ot = scaler.scale_[target_idx]
pred_ot = all_pred[:, :, target_idx] * scale_ot + mean_ot
true_ot = all_y[:, :, target_idx] * scale_ot + mean_ot
mse_ot = np.mean((pred_ot - true_ot) ** 2)
mae_ot = np.mean(np.abs(pred_ot - true_ot))
bias_ot = np.mean(pred_ot - true_ot)

print("\n==================== РЕЗУЛЬТАТЫ ====================")
print(f"Тестовый MSE (OT, исходный масштаб): {mse_ot:.6f}")
print(f"Тестовый MAE (OT, исходный масштаб): {mae_ot:.6f}")
print(f"Среднее смещение (pred - true)     : {bias_ot:.6f}")
print("===================================================")

# ============================================================================
# БЛОК 9: Визуализация
# ============================================================================
train_ot = train_data[:, target_idx]
val_ot = val_data[:, target_idx]
test_ot = test_data[:, target_idx]
test_len = len(test_ot)

# Усреднённый прогноз на тесте
pred_mean = np.full(test_len, np.nan)
for t in range(test_len):
    preds_at_t = []
    for i in range(len(all_pred)):
        start = i
        if start <= t < start + pred_len:
            preds_at_t.append(all_pred[i, t - start, target_idx] * scale_ot + mean_ot)
    if preds_at_t:
        pred_mean[t] = np.mean(preds_at_t)

total_len_full = len(train_ot) + len(val_ot) + len(test_ot)
train_idx = np.arange(0, len(train_ot))
val_idx = np.arange(len(train_ot), len(train_ot) + len(val_ot))
test_idx = np.arange(len(train_ot) + len(val_ot), total_len_full)

plt.figure(figsize=(16, 6))
plt.plot(train_idx, train_ot, label='Train (истинные)', color='blue', linewidth=1)
plt.plot(val_idx, val_ot, label='Val (истинные)', color='orange', linewidth=1)
plt.plot(test_idx, test_ot, label='Test (истинные)', color='green', linewidth=1.5)
plt.plot(test_idx, pred_mean, label='Прогноз (средний)', color='red', linestyle='--', linewidth=1.5)
plt.axvline(x=len(train_ot)-1, color='black', linestyle=':', linewidth=1, label='Граница Train/Val')
plt.axvline(x=len(train_ot)+len(val_ot)-1, color='black', linestyle=':', linewidth=1, label='Граница Val/Test')
plt.xlabel('Временной индекс')
plt.ylabel('OT (температура масла)')
plt.title('Весь ряд ETTh1: истинные и прогнозные значения')
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# 1:00

# ============================================================================
# БЛОК 1: Импорт библиотек
# ============================================================================
import torch
import torch.nn as nn
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from torch.utils.data import DataLoader, TensorDataset
from sklearn.preprocessing import StandardScaler
import warnings
warnings.filterwarnings('ignore')

# ============================================================================
# БЛОК 2: RevIN (Reversible Instance Normalization)
# ============================================================================
class RevIN(nn.Module):
    """
    Нормализация каждого канала по среднему и std внутри окна.
    Применяется перед входом в модель и инвертируется после прогноза.
    """
    def __init__(self, num_features, eps=1e-5):
        super().__init__()
        self.num_features = num_features
        self.eps = eps
        self.mean = None
        self.std = None

    def forward(self, x, mode='norm'):
        """
        x: [batch, n_vars, seq_len] или [batch, n_vars, pred_len]
        mode: 'norm' или 'denorm'
        """
        if mode == 'norm':
            # Вычисляем среднее и std по оси времени (последняя ось)
            self.mean = x.mean(dim=-1, keepdim=True)  # [batch, n_vars, 1]
            self.std = x.std(dim=-1, keepdim=True, unbiased=False) + self.eps
            x_norm = (x - self.mean) / self.std
            return x_norm
        elif mode == 'denorm':
            # Восстанавливаем исходный масштаб
            x_denorm = x * self.std + self.mean
            return x_denorm
        else:
            raise NotImplementedError

# ============================================================================
# БЛОК 3: Модель PatchTST с RevIN
# ============================================================================
class PatchTST(nn.Module):
    def __init__(self, seq_len, pred_len, n_vars, patch_len=16, stride=8,
                 d_model=128, n_heads=16, e_layers=3, dropout=0.2):
        super(PatchTST, self).__init__()
        self.seq_len = seq_len
        self.pred_len = pred_len
        self.n_vars = n_vars
        self.patch_len = patch_len
        self.stride = stride
        self.d_model = d_model

        # RevIN
        self.revin = RevIN(n_vars)

        # Количество патчей
        self.n_patches = int((seq_len - patch_len) / stride) + 1

        # Эмбеддинг патчей
        self.patch_embed = nn.Linear(patch_len, d_model)

        # Позиционное кодирование (обучаемое)
        self.pos_enc = nn.Parameter(torch.randn(1, self.n_patches, d_model))

        # Нормализация
        self.norm = nn.LayerNorm(d_model)

        # Трансформерный энкодер
        encoder_layer = nn.TransformerEncoderLayer(
            d_model=d_model,
            nhead=n_heads,
            dim_feedforward=d_model * 4,
            dropout=dropout,
            activation='gelu',
            batch_first=True
        )
        self.transformer = nn.TransformerEncoder(encoder_layer, num_layers=e_layers)

        # Выходной слой (голова)
        self.head = nn.Linear(self.n_patches * d_model, pred_len)

    def forward(self, x):
        """
        x: [batch, n_vars, seq_len]
        out: [batch, pred_len, n_vars]
        """
        batch_size = x.shape[0]

        # ----- RevIN (нормализация) -----
        x = self.revin(x, mode='norm')   # [batch, n_vars, seq_len]

        # ----- Channel‑independence -----
        x = x.permute(0, 2, 1)           # [batch, seq_len, n_vars]
        x = x.reshape(batch_size * self.n_vars, self.seq_len)  # [B*C, seq_len]

        # ----- Патчинг -----
        patches = x.unfold(dimension=1, size=self.patch_len, step=self.stride)  # [B*C, n_patches, patch_len]

        # ----- Эмбеддинг и позиционирование -----
        patches = self.patch_embed(patches)          # [B*C, n_patches, d_model]
        patches = patches + self.pos_enc             # добавляем позицию
        patches = self.norm(patches)                 # LayerNorm

        # ----- Трансформер -----
        encoded = self.transformer(patches)          # [B*C, n_patches, d_model]

        # ----- Голова (прогноз) -----
        encoded_flat = encoded.reshape(batch_size * self.n_vars, -1)  # [B*C, n_patches*d_model]
        pred = self.head(encoded_flat)               # [B*C, pred_len]

        # ----- Восстановление размерности -----
        pred = pred.reshape(batch_size, self.n_vars, self.pred_len)  # [batch, n_vars, pred_len]

        # ----- Инвертирование RevIN -----
        pred = self.revin(pred, mode='denorm')       # [batch, n_vars, pred_len]

        # Приводим к формату (batch, pred_len, n_vars)
        pred = pred.permute(0, 2, 1)
        return pred

# ============================================================================
# БЛОК 4: Гиперпараметры (увеличенная ёмкость)
# ============================================================================
SEQ_LEN = 96
PRED_LEN = 96
PATCH_LEN = 16
STRIDE = 8
D_MODEL = 128          # увеличено с 64
N_HEADS = 16           # увеличено с 8
E_LAYERS = 3           # увеличено с 2
DROPOUT = 0.4
BATCH_SIZE = 8
EPOCHS = 100           # увеличено
LR = 1e-4
PATIENCE = 10          # увеличено для надёжности
WEIGHT_DECAY = 1e-5
TRAIN_RATIO = 0.7
VAL_RATIO = 0.15
FILE_PATH = './ETTh1.csv'

# Устройство
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Используемое устройство: {device}")
if device.type == 'cuda':
    print(f"GPU: {torch.cuda.get_device_name(0)}")
    torch.backends.cudnn.benchmark = True

# ============================================================================
# БЛОК 5: Загрузка и подготовка данных
# ============================================================================
print("Загрузка данных...")
df = pd.read_csv(FILE_PATH, parse_dates=['date'], index_col='date')
features = ['HUFL', 'HULL', 'MUFL', 'MULL', 'LUFL', 'LULL', 'OT']
data = df[features].values
total_len = len(data)
print(f"Всего точек: {total_len}")

# Разбиение (по порядку, без перемешивания)
train_border = int(total_len * TRAIN_RATIO)
val_border = int(total_len * (TRAIN_RATIO + VAL_RATIO))

train_data = data[:train_border]
val_data = data[train_border:val_border]
test_data = data[val_border:]

print(f"Train: {len(train_data)} ({100*len(train_data)/total_len:.1f}%)")
print(f"Val:   {len(val_data)} ({100*len(val_data)/total_len:.1f}%)")
print(f"Test:  {len(test_data)} ({100*len(test_data)/total_len:.1f}%)")

# Стандартизация (масштабирование по обучающей выборке)
scaler = StandardScaler()
train_scaled = scaler.fit_transform(train_data)
val_scaled = scaler.transform(val_data)
test_scaled = scaler.transform(test_data)

# Создание скользящих окон
def create_sequences(data, seq_len, pred_len):
    xs, ys = [], []
    for i in range(len(data) - seq_len - pred_len + 1):
        x = data[i:i+seq_len]
        y = data[i+seq_len:i+seq_len+pred_len]
        xs.append(x)
        ys.append(y)
    return np.array(xs), np.array(ys)

X_train, y_train = create_sequences(train_scaled, SEQ_LEN, PRED_LEN)
X_val, y_val = create_sequences(val_scaled, SEQ_LEN, PRED_LEN)
X_test, y_test = create_sequences(test_scaled, SEQ_LEN, PRED_LEN)

print(f"Окон train: {X_train.shape[0]}, val: {X_val.shape[0]}, test: {X_test.shape[0]}")

# Датасеты и загрузчики
train_dataset = TensorDataset(torch.FloatTensor(X_train), torch.FloatTensor(y_train))
val_dataset = TensorDataset(torch.FloatTensor(X_val), torch.FloatTensor(y_val))
test_dataset = TensorDataset(torch.FloatTensor(X_test), torch.FloatTensor(y_test))

train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=BATCH_SIZE, shuffle=False)
test_loader = DataLoader(test_dataset, batch_size=BATCH_SIZE, shuffle=False)

n_vars = len(features)
target_idx = features.index('OT')
print(f"Число переменных: {n_vars}, индекс OT: {target_idx}")

# ============================================================================
# БЛОК 6: Инициализация модели, оптимизатор, планировщик, функция потерь
# ============================================================================
model = PatchTST(
    seq_len=SEQ_LEN, pred_len=PRED_LEN, n_vars=n_vars,
    patch_len=PATCH_LEN, stride=STRIDE,
    d_model=D_MODEL, n_heads=N_HEADS, e_layers=E_LAYERS,
    dropout=DROPOUT
).to(device)

print(f"Число параметров модели: {sum(p.numel() for p in model.parameters()):,}")

optimizer = torch.optim.Adam(model.parameters(), lr=LR, weight_decay=WEIGHT_DECAY)
criterion = nn.MSELoss()
scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer, mode='min', factor=0.5, patience=5, verbose=True
)

# ============================================================================
# БЛОК 7: Обучение с логированием потерь
# ============================================================================
print("\nОбучение начато...")
best_val_loss = float('inf')
counter = 0
train_losses, val_losses = [], []

for epoch in range(1, EPOCHS + 1):
    # --- Обучение ---
    model.train()
    train_loss = 0.0
    for X_batch, y_batch in train_loader:
        X_batch, y_batch = X_batch.to(device), y_batch.to(device)
        X_batch = X_batch.permute(0, 2, 1)       # [batch, seq_len, n_vars] -> [batch, n_vars, seq_len]
        optimizer.zero_grad()
        pred = model(X_batch)                    # [batch, pred_len, n_vars]
        loss = criterion(pred, y_batch)
        loss.backward()
        optimizer.step()
        train_loss += loss.item()
    train_loss /= len(train_loader)
    train_losses.append(train_loss)

    # --- Валидация ---
    model.eval()
    val_loss = 0.0
    with torch.no_grad():
        for X_batch, y_batch in val_loader:
            X_batch, y_batch = X_batch.to(device), y_batch.to(device)
            X_batch = X_batch.permute(0, 2, 1)
            pred = model(X_batch)
            loss = criterion(pred, y_batch)
            val_loss += loss.item()
    val_loss /= len(val_loader)
    val_losses.append(val_loss)

    # Планировщик
    scheduler.step(val_loss)

    # Early stopping и сохранение лучшей модели
    if val_loss < best_val_loss:
        best_val_loss = val_loss
        counter = 0
        torch.save(model.state_dict(), 'best_model.pth')
    else:
        counter += 1
        if counter >= PATIENCE:
            print(f"Ранняя остановка на эпохе {epoch}")
            break

    if epoch % 10 == 0 or epoch == 1:
        print(f"Эпоха {epoch:3d}/{EPOCHS} | Train Loss: {train_loss:.6f} | Val Loss: {val_loss:.6f}")

print("\nОбучение завершено. Загружаем лучшую модель...")
model.load_state_dict(torch.load('best_model.pth', weights_only=False))

# ============================================================================
# БЛОК 8: Тестирование и расчёт метрик
# ============================================================================
model.eval()
all_pred, all_y = [], []
with torch.no_grad():
    for X_batch, y_batch in test_loader:
        X_batch, y_batch = X_batch.to(device), y_batch.to(device)
        X_batch = X_batch.permute(0, 2, 1)
        pred = model(X_batch)
        all_pred.append(pred.cpu().numpy())
        all_y.append(y_batch.cpu().numpy())

all_pred = np.concatenate(all_pred, axis=0)   # (total_samples, pred_len, n_vars)
all_y = np.concatenate(all_y, axis=0)

# Извлекаем OT и переводим в исходный масштаб
mean_ot = scaler.mean_[target_idx]
scale_ot = scaler.scale_[target_idx]
pred_ot = all_pred[:, :, target_idx] * scale_ot + mean_ot
true_ot = all_y[:, :, target_idx] * scale_ot + mean_ot

# Метрики
mse = np.mean((pred_ot - true_ot) ** 2)
mae = np.mean(np.abs(pred_ot - true_ot))
bias = np.mean(pred_ot - true_ot)

print("\n==================== РЕЗУЛЬТАТЫ ====================")
print(f"Тестовый MSE (OT, исходный масштаб): {mse:.4f}")
print(f"Тестовый MAE (OT, исходный масштаб): {mae:.4f}")
print(f"Среднее смещение (pred - true)     : {bias:.4f}")
print(f"Станд. отклонение OT (train): {np.std(train_data[:, target_idx]):.2f} °C")
print(f"Диапазон OT (train): [{np.min(train_data[:, target_idx]):.1f}, {np.max(train_data[:, target_idx]):.1f}] °C")
print("===================================================")

# ============================================================================
# БЛОК 9: Бейзлайны (наивный и сезонный)
# ============================================================================
# Наивный прогноз (последнее значение окна)
naive_pred = all_y[:, -1, target_idx] * scale_ot + mean_ot  # (total_samples, pred_len?) – не совсем корректно
# Правильно: для каждого окна берём последнее значение входа, а не выходного окна
# Для этого нужно восстановить входы из test_loader, но проще использовать последнее значение каждого окна из X_test
# Достанем X_test из данных:
X_test_orig = test_scaled  # массив, который мы использовали для создания окон (но он не содержит окна, а содержит ряд)
# Поэтому переделаем: используем последнее значение каждого окна из all_x (если сохранили)
# Мы не сохраняли X, поэтому упростим: возьмём последнее значение из каждого окна в X_test (из тестового набора)
# Создадим X_test из последовательностей для бейзлайна
X_test_windows = X_test  # у нас уже есть X_test (окна) из create_sequences
last_val_naive = X_test[:, -1, target_idx] * scale_ot + mean_ot  # последнее значение в каждом окне (перед прогнозом)
# Теперь повторим это значение на pred_len шагов
naive_pred_full = np.repeat(last_val_naive.reshape(-1, 1), PRED_LEN, axis=1)

# Сезонный наивный (24 часа назад)
last_val_seasonal = X_test[:, -24, target_idx] * scale_ot + mean_ot  # значение 24 часа назад
seasonal_pred_full = np.repeat(last_val_seasonal.reshape(-1, 1), PRED_LEN, axis=1)

mse_naive = np.mean((naive_pred_full - true_ot) ** 2)
mae_naive = np.mean(np.abs(naive_pred_full - true_ot))
mse_seasonal = np.mean((seasonal_pred_full - true_ot) ** 2)
mae_seasonal = np.mean(np.abs(seasonal_pred_full - true_ot))

print("\n==================== БЕЙЗЛАЙНЫ ====================")
print(f"Наивный прогноз  -> MSE: {mse_naive:.4f}, MAE: {mae_naive:.4f}")
print(f"Сезонный (24ч)   -> MSE: {mse_seasonal:.4f}, MAE: {mae_seasonal:.4f}")
print(f"PatchTST          -> MSE: {mse:.4f}, MAE: {mae:.4f}")
print("===================================================")

# ============================================================================
# БЛОК 10: Визуализация
# ============================================================================
# 10.1 Кривые обучения
plt.figure(figsize=(12,4))
plt.subplot(1,2,1)
plt.plot(train_losses, label='Train Loss')
plt.plot(val_losses, label='Val Loss')
plt.xlabel('Эпоха')
plt.ylabel('MSE Loss')
plt.legend()
plt.grid()
plt.title('Кривые обучения')

# 10.2 Гистограмма ошибок
errors = (pred_ot - true_ot).flatten()
plt.subplot(1,2,2)
plt.hist(errors, bins=50, alpha=0.7, color='skyblue', edgecolor='black')
plt.axvline(x=0, color='red', linestyle='--', label='Нулевая ошибка')
plt.axvline(x=np.mean(errors), color='orange', linestyle='-', label=f'Среднее = {np.mean(errors):.2f}')
plt.xlabel('Ошибка (прогноз - истина)')
plt.ylabel('Частота')
plt.legend()
plt.grid()
plt.title(f'Распределение ошибок (bias={np.mean(errors):.2f} °C)')
plt.tight_layout()
plt.show()

# 10.3 Примеры прогнозов (3 случайных окна)
indices = np.random.choice(len(all_pred), 3, replace=False)
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for i, idx in enumerate(indices):
    axes[i].plot(true_ot[idx], label='Истина', color='green', linewidth=2)
    axes[i].plot(pred_ot[idx], label='Прогноз', color='red', linestyle='--', linewidth=2)
    axes[i].set_title(f'Окно {idx}')
    axes[i].set_xlabel('Шаг прогноза')
    axes[i].set_ylabel('OT (°C)')
    axes[i].legend()
    axes[i].grid()
plt.tight_layout()
plt.show()

# 10.4 Scatter plot
plt.figure(figsize=(6,6))
plt.scatter(true_ot.flatten(), pred_ot.flatten(), s=1, alpha=0.5, color='blue')
min_val = min(true_ot.min(), pred_ot.min())
max_val = max(true_ot.max(), pred_ot.max())
plt.plot([min_val, max_val], [min_val, max_val], 'k--', linewidth=1, label='Идеальная линия')
plt.xlabel('Истинные значения (°C)')
plt.ylabel('Предсказанные значения (°C)')
plt.title(f'Scatter plot (MSE={mse:.2f}, MAE={mae:.2f})')
plt.legend()
plt.grid()
plt.show()

# 10.5 Весь ряд с прогнозом (усреднённый)
train_ot = train_data[:, target_idx]
val_ot = val_data[:, target_idx]
test_ot = test_data[:, target_idx]
test_len = len(test_ot)

# Усреднённый прогноз на тесте (среднее по всем окнам, покрывающим момент)
pred_mean = np.full(test_len, np.nan)
for t in range(test_len):
    preds_at_t = []
    for i in range(len(all_pred)):
        start = i  # каждое окно начинается с индекса i в тестовом ряду
        if start <= t < start + PRED_LEN:
            preds_at_t.append(all_pred[i, t - start, target_idx] * scale_ot + mean_ot)
    if preds_at_t:
        pred_mean[t] = np.mean(preds_at_t)

total_len_full = len(train_ot) + len(val_ot) + len(test_ot)
train_idx = np.arange(0, len(train_ot))
val_idx = np.arange(len(train_ot), len(train_ot) + len(val_ot))
test_idx = np.arange(len(train_ot) + len(val_ot), total_len_full)

plt.figure(figsize=(16,6))
plt.plot(train_idx, train_ot, label='Train', color='blue', linewidth=1)
plt.plot(val_idx, val_ot, label='Val', color='orange', linewidth=1)
plt.plot(test_idx, test_ot, label='Test (истина)', color='green', linewidth=1.5)
plt.plot(test_idx, pred_mean, label='Прогноз (средний)', color='red', linestyle='--', linewidth=1.5)
plt.axvline(x=len(train_ot)-1, color='black', linestyle=':', linewidth=1)
plt.axvline(x=len(train_ot)+len(val_ot)-1, color='black', linestyle=':', linewidth=1)
plt.xlabel('Временной индекс')
plt.ylabel('OT (°C)')
plt.title('Весь ряд ETTh1: прогноз на тестовой части')
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

print("\nВсе графики построены.")
```
